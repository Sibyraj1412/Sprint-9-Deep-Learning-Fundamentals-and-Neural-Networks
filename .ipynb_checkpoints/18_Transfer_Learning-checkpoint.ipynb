# Notebook 18: Introduction to Transfer Learning

**Transfer learning** reuses representations learned on a large source dataset for a related target task. Instead of learning every visual feature from random initialization, we start with an open-source TorchVision ImageNet model and adapt its classifier to Fashion-MNIST.

Workflow: **Pretrained model → freeze or modify layers → train on the new dataset → optionally fine-tune**. Pretrained weights require an internet connection on first download; the notebook has a random-weight fallback for API/code practice, but that fallback is not pretrained transfer learning.

## 1. Concepts and when to use them
- **Pretrained model:** architecture plus weights learned on a source task (here, ImageNet).
- **Feature extraction:** freeze the backbone and train only a new task-specific output head. Frozen parameters do not receive gradients, reducing compute and preserving learned features.
- **Fine-tuning:** unfreeze some or all pretrained layers and continue training with a small learning rate. This adapts representations to the target domain but can overfit or damage useful weights if the learning rate is too high.
- **Trainable/frozen layers:** controlled through each parameter's `requires_grad` flag.

Transfer learning is especially useful when labeled target data or compute is limited and source/target images are reasonably related. Training from scratch can make sense with abundant data, a very different domain, or a specialized architecture. Fashion-MNIST is grayscale and low-resolution, unlike ImageNet; this is a useful caveat when interpreting results.

```python
# If required in the active notebook kernel: %pip install torch torchvision
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms, models

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Device:', device)
weights = models.ResNet18_Weights.DEFAULT
try:
    model = models.resnet18(weights=weights)
    using_pretrained = True
    print('Loaded ImageNet pretrained ResNet-18 weights.')
except Exception as error:
    print(f'Could not download/load pretrained weights ({type(error).__name__}: {error}). Using random weights for demonstration only.')
    model = models.resnet18(weights=None)
    using_pretrained = False

# Replace the original 1000-class ImageNet head with 10 Fashion-MNIST classes.
input_features = model.fc.in_features
model.fc = nn.Linear(input_features, 10)
for parameter in model.parameters(): parameter.requires_grad = False
for parameter in model.fc.parameters(): parameter.requires_grad = True
model = model.to(device)
print('Trainable parameter count:', sum(p.numel() for p in model.parameters() if p.requires_grad))
```

## 2. Prepare target images
ResNet expects three-channel images. Fashion-MNIST is grayscale, so we replicate grayscale across RGB channels, resize to 96×96 to reduce compute, and normalize with ImageNet statistics. The first 1,200 training examples and 300 test examples are used as a small demonstration subset; use larger subsets and a separate validation split for serious evaluation.

```python
transform = transforms.Compose([transforms.Resize((96, 96)), transforms.Grayscale(num_output_channels=3), transforms.ToTensor(),
                               transforms.Normalize((0.485, 0.456, 0.406), (0.229, 0.224, 0.225))])
train_full = datasets.FashionMNIST('./data', train=True, download=True, transform=transform)
test_full = datasets.FashionMNIST('./data', train=False, download=True, transform=transform)
train_ds = Subset(train_full, range(1200)); test_ds = Subset(test_full, range(300))
train_loader = DataLoader(train_ds, batch_size=32, shuffle=True, num_workers=0)
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False, num_workers=0)
print('Train subset:', len(train_ds), '| test subset:', len(test_ds))
```

## 3. Feature extraction: train the new head
Only the new fully connected head is trainable. The pretrained convolutional backbone acts as a fixed feature extractor. In the random-weight fallback, freezing the backbone is not meaningful; set `requires_grad=True` for all parameters or rerun when pretrained weights can be loaded.

```python
if not using_pretrained:
    for parameter in model.parameters(): parameter.requires_grad = True
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam((p for p in model.parameters() if p.requires_grad), lr=1e-3)

for epoch in range(1):
    model.train(); total_loss = 0.0
    for images, labels in train_loader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad(); loss = criterion(model(images), labels); loss.backward(); optimizer.step()
        total_loss += loss.item() * labels.size(0)
    print(f'Epoch {epoch+1}: mean training loss={total_loss / len(train_ds):.4f}')

model.eval(); correct = 0; count = 0
with torch.no_grad():
    for images, labels in test_loader:
        images, labels = images.to(device), labels.to(device)
        correct += (model(images).argmax(1) == labels).sum().item(); count += labels.size(0)
print(f'Demo-subset test accuracy: {correct / count:.3f}')
```

## 4. Fine-tuning selected layers
A common next step is to unfreeze only the final backbone block and train it with a smaller learning rate than the classifier. This cell demonstrates the parameter and optimizer setup; run it after feature extraction if you want to continue training. For best practice, add validation monitoring and save the best validation checkpoint.

```python
# Unfreeze the final residual block while keeping earlier features fixed.
for parameter in model.layer4.parameters(): parameter.requires_grad = True
fine_tune_optimizer = torch.optim.Adam([
    {'params': model.layer4.parameters(), 'lr': 1e-4},
    {'params': model.fc.parameters(), 'lr': 3e-4}
], weight_decay=1e-4)
```
